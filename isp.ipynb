```python
dados = "https://www.ispdados.rj.gov.br/Arquivos/BaseDPEvolucaoMensalCisp.csv"
df = pd.read_csv(dados, encoding='latin-1', sep=';')
df
```

```python
# Agrupando por CISP e somando os furtos de celular
df_cisp = df.groupby('cisp')['furto_celular'].sum().reset_index()
# Ordenando do maior para o menor
df_cisp = df_cisp.sort_values(by='furto_celular', ascending=False)
df_cisp
```

```python
df_capital = df.loc[df['munic'] == 'Rio de Janeiro']
df_furto_veiculos_cisp_capital = df_capital.groupby('cisp')['furto_veiculos'].sum().reset_index().sort_values(by='furto_veiculos', ascending=False)
df_furto_veiculos_cisp_capital
```

```python
df_furto_veiculos_cisp_capital.tail(10)
```

```python
df_cisp_furto_capital = df.loc[df['munic'] == 'Rio de Janeiro']
df_furto_veiculos_cisp_capital = df_cisp_furto_capital.groupby('cisp')['furto_veiculos'].sum().reset_index().sort_values(by='furto_veiculos', ascending=False)
df_furto_veiculos_cisp_capital
```

```python
array_furto_veiculos_cisp_capital= np.array(df_furto_veiculos_cisp_capital['furto_veiculos'])
array_furto_veiculos_cisp_capital
```

```python
media = np.mean(array_furto_veiculos_cisp_capital)
mediana = np.median(array_furto_veiculos_cisp_capital)
print(f'Média {round(media,2)}\nMediana {mediana}')
```

```python
q1 = np.percentile(array_furto_veiculos_cisp_capital, 25)
q2 = np.percentile(array_furto_veiculos_cisp_capital, 50)
q3 = np.percentile(array_furto_veiculos_cisp_capital, 75)
print('Q1:', q1)
print('Q2:', q2)
print('Q3:', q3)
```

```python
top_25= df_furto_veiculos_cisp_capital.loc[df_furto_veiculos_cisp_capital['furto_veiculos']>q3]
top_25
```
