```python
import pandas as pd
import numpy as np
```

```python
df_vendagasolina = df[df['Produto'] == 'GASOLINA']
df_vendagasolinarj = df_vendagasolina[df_vendagasolina['Estado - Sigla'] == 'RJ']
df_vendagasolinarj
 
```

```python
array_vendagasolinarj = np.array(df_vendagasolinarj['Valor de Venda'])
media = np.mean(array_vendagasolinarj)
mediana = np.median(array_vendagasolinarj)
dist = (media-mediana)/mediana
q1 =np.percentile(array_vendagasolinarj,25)
q3 =np.percentile(array_vendagasolinarj,75)
iqr = q3 - q1
limite_superior = q3 +(1.5*iqr)
limite_inferior = q1 - (1.5*iqr)

print(f"Média: {media:.2f}")
print(f"Mediana: {mediana:.2f}")
print(f"Distância: {dist:.2%}")
print(f"q1: {q1}")
print(f"q3: {q3}")
print(f"Intervalo interquartil: {iqr:.2%}")
print(f"Limite Superior: {limite_superior:.2f}")
print(f"Limite Inferior: {limite_inferior:.2f}")
```

```python
#Sim, alguns poucos postos praticam médias de valores acima do padrão porém o valor não é extremamente fora do padrão praticado.
```

```python
outlier_vendagasolinarjsuperior = df_vendagasolinarj.loc[df_vendagasolinarj['Valor de Venda']>limite_superior]
outlier_vendagasolinarjsuperior.head(3)
```

```python
outlier_vendagasolinarjsuperior = df_vendagasolinarj.loc[df_vendagasolinarj['Valor de Venda']>limite_inferior]
outlier_vendagasolinarjsuperior.tail(3)
```

```python
#O limite inferior não é negativo, logo, existe um valor limite mínimo praticado de venda
```

```python
df_vendagasolina = df[df['Produto'] == 'GASOLINA']
df_vendagasolina_mun_rj = df_vendagasolina[df_vendagasolina['Municipio'] == 'RIO DE JANEIRO']
df_vendagasolina_mun_rj
```

```python
df_vendagasolina_bairro_rj = df_vendagasolina_mun_rj.groupby('Bairro')['Valor de Venda'].mean().reset_index().sort_values(by='Valor de Venda', ascending=False)
df_vendagasolina_bairro_rj
```

```python
array_vendagasolina_bairro_rj = np.array(df_vendagasolina_bairro_rj['Valor de Venda'])
media_gaso_rj = np.mean(array_vendagasolina_bairro_rj)
mediana_gaso_rj = np.median(array_vendagasolina_bairro_rj)
dist_gaso_rj = (media_gaso_rj-mediana_gaso_rj)/mediana_gaso_rj
q1_gaso_rj =np.percentile(array_vendagasolina_bairro_rj,25)
q3_gaso_rj =np.percentile(array_vendagasolina_bairro_rj,75)
iqr_gaso_rj = q3_gaso_rj - q1_gaso_rj
limite_superior_gaso_rj = q3_gaso_rj +(1.5*iqr_gaso_rj)
limite_inferior_gaso_rj = q1_gaso_rj - (1.5*iqr_gaso_rj)

print(f"Média: {media_gaso_rj:.2f}")
print(f"Mediana: {mediana_gaso_rj:.2f}")
print(f"Distância: {dist_gaso_rj:.2%}")
print(f"q1: {q1_gaso_rj}")
print(f"q3: {q3_gaso_rj}")
print(f"Intervalo interquartil: {iqr_gaso_rj:.2%}")
print(f"Limite Superior: {limite_superior_gaso_rj:.2f}")
print(f"Limite Inferior: {limite_inferior_gaso_rj:.2f}")
```
